```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data = {
    'Name': ['Asha', 'Bharath', 'Chetan', 'Divya', 'Esha'],
    'Age': [20, 21, None, 22, 23],
    'Marks': [85, None, 78, 90, None],
    'City': ['Mysuru', 'Bengaluru', 'Mysuru', None, 'Mandya']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Mean imputation
df_mean = df.copy()
df_mean['Age'] = df_mean['Age'].fillna(df_mean['Age'].mean())

print("\nAfter Mean Imputation:")
print(df_mean)

# Median imputation
df_median = df.copy()
df_median['Marks'] = df_median['Marks'].fillna(df_median['Marks'].median())

print("\nAfter Median Imputation:")
print(df_median)

# Forward fill
df_ffill = df.copy()
df_ffill['City'] = df_ffill['City'].ffill()

print("\nAfter Forward Fill:")
print(df_ffill)

# Categorical encoding
df_encoded = df_ffill.copy()
encoder = LabelEncoder()
df_encoded['City'] = encoder.fit_transform(df_encoded['City'])

print("\nAfter Categorical Encoding:")
print(df_encoded)
```
